```python
import tensorflow as tf
from tensorflow.keras.preprocessing import image
import numpy as np
import os

# --- 1. LOAD THE TRAINED MODEL ---
model_path = 'models/corn_disease_classifier.h5'
print(f"Loading model from: {model_path}")
model = tf.keras.models.load_model(model_path)
print("Model loaded successfully.")

# --- 2. PREPARE A TEST IMAGE ---

# IMPORTANT: Find an image of a corn leaf on the internet (or from your dataset)
# and save it directly inside your 'ai_model' folder.
# Then, change the 'your_test_image.jpg' string below to the actual filename.

image_filename = 'test_model3.png' # <--- CHANGE THIS FILENAME

# Check if the image exists
if not os.path.exists(image_filename):
    print(f"ERROR: The file '{image_filename}' was not found in the 'ai_model' directory.")
    print("Please download a test image, place it in the 'ai_model' folder, and update the filename in this script.")
else:
    # Load and preprocess the image
    img_size = (128, 128)
    img = image.load_img(image_filename, target_size=img_size)
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0) # Create a batch
    img_array /= 255.0 # Rescale to 0-1

    # --- 3. MAKE A PREDICTION ---
    prediction = model.predict(img_array)

    # --- 4. INTERPRET THE RESULT ---
    # The output of a sigmoid is a value between 0 and 1.
    # By convention, 'diseased' is often class 0 and 'healthy' is class 1.
    
    print(f"\nRaw prediction value: {prediction[0][0]}")
    
    if prediction[0][0] > 0.5:
        print("Prediction: This is a HEALTHY corn leaf.")
    else:
        print("Prediction: This is a DISEASED corn leaf.")
```
